# 03 — Composition over time

Yearly trends in GC content and CpG O/E (2015–2026), including the COVID-19 disruption (2020–2021).

```python
import sys, os
sys.path.insert(0, os.path.abspath("../src"))
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style="whitegrid")
pd.set_option("display.width", 120)
from rsv import io, composition
```

```python
d = pd.read_csv("../data/processed/rsv_features.tsv", sep="\t")
d = d[(d.year>=2015)&(d.year<=2026)]
yr = d.groupby(["Subtype","year"]).agg(n=("gc_content","size"),
      gc=("gc_content","mean"), oe_CG=("oe_CG","mean")).reset_index()
yr = yr[yr.n>=20]
yr.head()
```

### Trend correlation (mean composition vs year)

```python
from scipy.stats import spearmanr
for s in "AB":
    x=yr[yr.Subtype==s]
    for f in ["gc","oe_CG"]:
        r,p = spearmanr(x.year, x[f])
        print(f"{s} {f:6s} rho={r:+.3f} p={p:.3f}")
```

```python
fig,ax=plt.subplots(1,2,figsize=(12,4))
for s,c in zip("AB",["#2c7fb8","#de2d26"]):
    x=yr[yr.Subtype==s]
    ax[0].plot(x.year,x.gc,"-o",color=c,label=s)
    ax[1].plot(x.year,x.oe_CG,"-o",color=c,label=s)
ax[0].set_title("mean GC content"); ax[1].set_title("mean CpG O/E")
ax[0].axvspan(2020,2021,color="grey",alpha=.15)
ax[1].axvspan(2020,2021,color="grey",alpha=.15)
ax[0].legend(title="subtype"); plt.show()
```

Grey band = COVID-19 disruption. Note A and B diverge in CpG O/E after 2021.
